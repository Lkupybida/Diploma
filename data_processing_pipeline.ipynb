```python
print(f"Stations before dropping: {len(df['eic_code'].unique())}, rows: {len(df)}")

global_end = df["datetime"].max()

# station-level stats
g = df.groupby("eic_code")["datetime"]

station_stats = pd.DataFrame({
    "start": g.min(),
    "end": g.max(),
    "n_hours": g.count()
})

# rule 1: must have data near dataset end (within 7 days tolerance)
station_stats["reaches_end"] = station_stats["end"] >= (global_end - pd.Timedelta(days=7))

# rule 2: must have at least 3 months of REAL observations
station_stats[">=3_months_data"] = station_stats["n_hours"] >= 90 * 24

# final filter
valid_stations = station_stats[
    station_stats["reaches_end"] &
    station_stats[">=3_months_data"]
].index

df = df[df["eic_code"].isin(valid_stations)].copy()

print(f"Stations after dropping: {len(df['eic_code'].unique())}, rows: {len(df)}")
```

```python
prices_df = pd.read_excel('data/additional_data/ВхідніДані.xlsx', sheet_name="Prices", header=1)
cols_to_rename = {
    'Row Labels': 'Date/Hour',
    'БР- (продаж)': 'selling_price',
    'БР+ (купівля)': 'buying price',
    'РДН': 'day_ahead_price'
}
prices_df = prices_df.rename(columns=cols_to_rename).copy()
prices_df
```

```python
# identify rows that contain full datetime
is_datetime = prices_df['Date/Hour'].astype(str).str.contains('-')

# extract base date
prices_df.loc[is_datetime, 'base_date'] = pd.to_datetime(prices_df.loc[is_datetime, 'Date/Hour'])

# forward fill the date
prices_df['base_date'] = prices_df['base_date'].ffill()

# extract hour rows
prices_df['hour'] = pd.to_numeric(prices_df['Date/Hour'], errors='coerce')

# build final datetime
prices_df['datetime'] = prices_df['base_date']

# where hour exists → add (hour-1)
mask = prices_df['hour'].notna()
prices_df.loc[mask, 'datetime'] = (
    prices_df.loc[mask, 'base_date'] +
    pd.to_timedelta(prices_df.loc[mask, 'hour'] - 1, unit='h')
)

# drop helper columns if you want
prices_df = prices_df.drop(columns=['base_date', 'hour', 'Date/Hour'])
prices_df.dropna(inplace=True)
prices_df
```

```python
full_range = pd.date_range(prices_df['datetime'].min(),
prices_df['datetime'].max(), freq='h')
prices_df = (
  prices_df
  .set_index('datetime')
  .reindex(full_range)
  .ffill()  # carry previous hour's price into the phantom hour
  .reset_index()
  .rename(columns={'index': 'datetime'})
)
```

```python
df = df.merge(prices_df, on='datetime', how='left', suffixes=('', '_dict'))
df
```
