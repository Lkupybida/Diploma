```python
import pandas as pd
import os
from tqdm.notebook import tqdm
```

```python
folder = "data/excel_files"

file_paths = [os.path.join(folder, f) for f in os.listdir(folder)]
file_paths
```

```python
df = pd.DataFrame()
for f in tqdm(file_paths):
    df = pd.concat([df, pd.read_excel(f, sheet_name="DATA", header=5)], ignore_index=True)

df = df[df['EIC-код'] != 'Grand Total'].copy()
df
```

```python
static_data_dict = pd.read_excel("data/raw/8month2025.xlsx", sheet_name="Dict")
static_data_dict
```

```python
static_data_dict['ОСР опис'].value_counts()
```

```python
df = df.merge(static_data_dict, on='EIC-код', how='left', suffixes=('', '_dict'))
df
```

```python
df['Дата'] = pd.to_datetime(df['Дата'], errors='coerce')
df['Дата'] = df['Дата'] + pd.to_timedelta(df['Hour'].fillna(0), unit='h')
df
```

```python
cols_to_leave = ['EIC-код', 'Група', 'Дата', 'Sum of кВт',
                 'Average of Ціна розподілу ЕЕ грн. без ПДВ/кВт*год', 'Average of Ціна ЕЕ грн. без ПДВ/кВт*год',
                 'Область', 'GPS-координати - Широта', 'GPS-координати - Довгота', 'ОСР опис']

rename_cols = {
    'Дата': 'datetime',
    'EIC-код': 'EIC-код_cat',
    'Група': 'Група_cat',
    'Average of Ціна розподілу ЕЕ грн. без ПДВ/кВт*год': 'Ціна розподілу ЕЕ',
    'Average of Ціна ЕЕ грн. без ПДВ/кВт*год': 'Ціна ЕЕ',
    'Область': 'Область_cat',
    'GPS-координати - Широта': 'Широта',
    'GPS-координати - Довгота': 'Довгота',
    'ОСР опис': 'ОСР опис_cat'
}

weather_cols_to_create = ['temperature_2m',
       'apparent_temperature', 'dew_point_2m', 'relative_humidity_2m',
       'precipitation', 'rain', 'snowfall', 'cloud_cover', 'cloud_cover_low',
       'cloud_cover_mid', 'cloud_cover_high', 'surface_pressure',
       'wind_speed_10m', 'wind_direction_10m', 'wind_gusts_10m',
       'shortwave_radiation', 'diffuse_radiation', 'direct_normal_irradiance']
```
