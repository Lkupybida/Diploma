```python
def generate_datetime_per_eic(
    df: pd.DataFrame,
    start: str,
    end: str,
    freq: str = "H"
) -> pd.DataFrame:
    """
    For each unique 'EIC-код_cat', generate a full datetime range
    and keep all other column values constant per EIC.

    Parameters:
    - df: input dataframe (must contain 'EIC-код_cat')
    - start: start datetime
    - end: end datetime
    - freq: frequency (default 'H')

    Returns:
    - Expanded DataFrame with datetime + original columns
    """

    # Ensure unique rows per EIC (important!)
    base = df.drop_duplicates(subset=['EIC-код_cat'])

    # Create datetime range
    date_range = pd.date_range(start=start, end=end, freq=freq)

    # Create cartesian product
    expanded = pd.MultiIndex.from_product(
        [base['EIC-код_cat'], date_range],
        names=['EIC-код_cat', 'datetime']
    ).to_frame(index=False)

    # Merge back original columns
    result = expanded.merge(base.drop(columns=['datetime']), on='EIC-код_cat', how='left')

    return result

future_df = generate_datetime_per_eic(df, start = '2025-09-01 00:00:00+0300', end = '2026-04-01 00:00:00+0300')
future_df
```

```python
WEATHER_FILE  = "../data/no_y_col/with_weather.parquet"
TARGET_FILE   = "../data/no_y_col/no_weather.parquet"
OUTPUT_FILE   = "../data/no_y_col/with_weather_v2.parquet"

# Column names used as the join key — adjust if your files use different names
LAT_COL      = "Широта"
LON_COL      = "Довгота"
DT_COL       = "datetime"

WEATHER_COLS = FUTURE_REALS

# ── Load ──────────────────────────────────────────────────────────────────────

print(f"Loading {WEATHER_FILE} ...")
weather_df = pd.read_parquet(WEATHER_FILE)
weather_df = weather_df.rename(columns={
    'GPS-координати - Широта': 'Широта',
    'GPS-координати - Довгота': 'Довгота'
})

print(f"Loading {TARGET_FILE} ...")
target_df  = pd.read_parquet(TARGET_FILE)

# ── Validate columns exist ────────────────────────────────────────────────────

key_cols = [LAT_COL, LON_COL, DT_COL]

for col in key_cols:
    assert col in weather_df.columns, f"Key column '{col}' missing from {WEATHER_FILE}"
    assert col in target_df.columns,  f"Key column '{col}' missing from {TARGET_FILE}"

missing_weather_cols = [c for c in WEATHER_COLS if c not in weather_df.columns]
if missing_weather_cols:
    raise ValueError(f"Weather columns missing from {WEATHER_FILE}: {missing_weather_cols}")

# ── Align datetime types (both files must match for the join to work) ─────────

for df, name in [(weather_df, WEATHER_FILE), (target_df, TARGET_FILE)]:
    if not pd.api.types.is_datetime64_any_dtype(df[DT_COL]):
        print(f"  Converting '{DT_COL}' in {name} to datetime ...")
        df[DT_COL] = pd.to_datetime(df[DT_COL])

# ── Build a lookup table from the weather file ────────────────────────────────
# Keep only the columns we need to avoid bloating memory on large files.

lookup = (
    weather_df[key_cols + WEATHER_COLS]
    .drop_duplicates(subset=key_cols)   # one row per (lat, lon, dt) combo
    .set_index(key_cols)
)

print(f"Lookup table: {len(lookup):,} unique (lat, lon, datetime) combos")

# ── Drop pre-existing weather columns from target (avoid _x/_y suffixes) ──────

cols_to_drop = [c for c in WEATHER_COLS if c in target_df.columns]
if cols_to_drop:
    print(f"  Dropping existing columns from target: {cols_to_drop}")
    target_df = target_df.drop(columns=cols_to_drop)

# ── Merge ─────────────────────────────────────────────────────────────────────

print("Merging weather columns ...")
result_df = target_df.merge(
    lookup.reset_index(),
    on=key_cols,
    how="left",       # keep all rows from target; unmatched rows → NaN
    validate="m:1",   # each target row matches at most one lookup row
)

# ── Report coverage ───────────────────────────────────────────────────────────

n_total   = len(result_df)
n_matched = result_df["temperature_2m"].notna().sum()
n_missing = n_total - n_matched

print(f"\nRows total  : {n_total:,}")
print(f"Rows matched: {n_matched:,}")
if n_missing:
    print(f"Rows with no weather match (NaN): {n_missing:,}")
    print("  → Check that lat/lon precision and datetime format are identical in both files.")
```
