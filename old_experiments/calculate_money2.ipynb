```python
df = pd.read_parquet("../data/silver/dataset_with_weather_time_features.parquet")
```

```python
df
```

```python
# categorical columns
categorical_cols = ['Група', 'АЗС', 'Тип', 'Область', 'ОСР код', 'ОСР опис', 'EIC-код']

cols_to_drop = ['Year', 'Month', 'Day', 'Hour', 'day_of_week', 'season_number', 'Адреса']

df = df.drop(columns=cols_to_drop)

# 1. Add _cat suffix to categorical columns
df = df.rename(columns={col: f"{col}_cat" for col in categorical_cols})

# 2. Rename price columns
df = df.rename(columns={
    'Average of Ціна розподілу ЕЕ грн. без ПДВ/кВт*год': 'Ціна розподілу ЕЕ',
    'Average of Ціна ЕЕ грн. без ПДВ/кВт*год': 'Ціна ЕЕ',
    'GPS-координати - Широта': 'Широта',
    'GPS-координати - Довгота': 'Довгота'
})
df
```

```python
df['Sum of кВт'] = df['Sum of кВт'].clip(lower=0)

df["Money_spent"] = (
    df["Sum of кВт"]
    * (
        df['Ціна розподілу ЕЕ']
        + df['Ціна ЕЕ']
      )
)
df
```

```python
test = df[df['datetime'] > '2025-07-31']
test.to_parquet("data/silver_money_calc/test.parquet", index=False)

val = df[(df['datetime'] > '2025-06-30') & (df['datetime'] < '2025-08-01')]
val.to_parquet("data/silver_money_calc/val.parquet", index=False)

train = df[df['datetime'] < '2025-07-01']
train.to_parquet("data/silver_money_calc/train.parquet", index=False)
```
