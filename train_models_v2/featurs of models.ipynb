```python
import torch

ckpt_path = "crossformer_best.pt"

# Load checkpoint (CPU-safe)
ckpt = torch.load(ckpt_path, map_location="cpu")

print("=== CHECKPOINT KEYS ===")
for k in ckpt.keys():
    print(k)

print("\n=== TRYING TO FIND FEATURES ===")

# Common places where features might be stored
possible_paths = [
    ("hparams", "features"),
    ("hparams", "input_features"),
    ("hparams", "x_cols"),
    ("hparams", "feature_names"),
    ("dataset", "features"),
    ("dataset", "input_features"),
    ("config", "features"),
]

found = False

for path in possible_paths:
    obj = ckpt
    try:
        for p in path:
            obj = obj[p]
        print(f"\nFound features at {'.'.join(path)}:")
        print(obj)
        found = True
    except (KeyError, TypeError):
        continue

if not found:
    print("No explicit feature list found. Dumping hparams/config if available:\n")

    if "hparams" in ckpt:
        print("=== hparams ===")
        for k, v in ckpt["hparams"].items():
            print(f"{k}: {v}")

    if "config" in ckpt:
        print("\n=== config ===")
        print(ckpt["config"])

# Optional: inspect model structure
if "state_dict" in ckpt:
    print("\n=== MODEL LAYERS ===")
    for k in list(ckpt["state_dict"].keys())[:20]:
        print(k)
```
